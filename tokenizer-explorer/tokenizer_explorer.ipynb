# Tokenizer Explorer

This notebook explores how different tokenizers convert text into tokens and token IDs.

We will:
- Count characters
- Count words
- Estimate token count
- Generate actual token IDs
- Decode token IDs back into text
- Compare GPT, WordPiece, and SentencePiece tokenization

```python
def analyze_text(text_b):
    character_count = len(text_b)
    word_count = len(text_b.split())
    token_ids = encoding.encode(text_b)
    token_count = len(token_ids)
    estimated_token_count = round(character_count/4,2)
    
    print('text',text_b)
    print()
    print('character count: ',character_count)
    print('word count: ',word_count)
    print('etc: ', estimated_token_count)
    print('actual token count: ', token_count)
```

```python
text_b = "ESP32 autofocus failed due to SPI timeout."

analyze_text(text_b)
```

```python
import pandas as pd
```

```python
def token_table(text):
    token_ids = encoding.encode(text)

    rows = []

    for position, token_id in enumerate(token_ids):
        token_bytes = encoding.decode_single_token_bytes(token_id)

        try:
            token_text = token_bytes.decode("utf-8")
        except UnicodeDecodeError:
            token_text = str(token_bytes)

        rows.append({
            "Position": position,
            "Token ID": token_id,
            "Token": token_text
        })

    return pd.DataFrame(rows)
```

```python
text = "ESP32 autofocus failed due to SPI timeout."

token_table(text)
```

```python
token_ids = encoding.encode(text)
decoded_text = encoding.decode(token_ids)

print('original: ',text)
print('decoded text: ',decoded_text)

print('same text: ', text == decoded_text)
```

# Experiment: GPT vs SentencePiece vs WordPiece

We use the exact same sentence with three different tokenizers:

> ESP32 autofocus failed due to SPI timeout.

The goal is to compare how each tokenizer splits the same text and how many tokens are produced.

```python
sentence = 'esp32 autofocus succeded with spi'
print(sentence)
```

```python
gpt_token_ids = encoding.encode(sentence)

print('gpt token ids: ',gpt_token_ids)
print('gpt token count: ', len(gpt_token_ids))
```

```python
gpt_tokens = []

for token_id in gpt_token_ids:
    token_bytes = encoding.decode_single_token_bytes(token_id)
    
    try:
        token_text = token_bytes.decode("utf-8")
    except UnicodeDecodeError:
        token_text = str(token_bytes)
    
    gpt_tokens.append(token_text)

print(gpt_tokens)
```

```python
from transformers import AutoTokenizer
```

```python
sp_tokenizer = AutoTokenizer.from_pretrained("google/mt5-small")
```

```python
x = sp_tokenizer(sentence, add_special_tokens = False)
 
sp_token_ids = x['input_ids']
print('sentence token ids: ', sp_token_ids)

print('sentencepiece token count: ',len(sp_token_ids))

```

```python
sp_tokens = sp_tokenizer.convert_ids_to_tokens(sp_token_ids)

print("SentencePiece Tokens:")
print(sp_tokens)
```

```python
sp_table = pd.DataFrame({
    "Position": range(len(sp_token_ids)),
    "Token ID": sp_token_ids,
    "Token": sp_tokens
})

sp_table
```

```python
sp_decoded = sp_tokenizer.decode(
    sp_token_ids,
    skip_special_tokens=True
)

print("Original:")
print(sentence)

print("\nDecoded:")
print(sp_decoded)

print("\nSame text:", sentence == sp_decoded)
```

```python
from tokenizers import Tokenizer
```

```python
wordpiece_tokenizer = Tokenizer.from_pretrained("bert-base-uncased")
```

```python
wp_output = wordpiece_tokenizer.encode(sentence)

wp_token_ids = wp_output.ids
wp_tokens = wp_output.tokens

print("WordPiece Token IDs for sentence:")
print(wp_token_ids)

print("\nWordPiece Token Count :")
print(len(wp_token_ids))

print('the sentence is: ',sentence)
```

```python
print("WordPiece Tokens:")
print(wp_tokens)
```

```python
wp_table = pd.DataFrame({
    "Position": range(len(wp_token_ids)),
    "Token ID": wp_token_ids,
    "Token": wp_tokens
})

wp_table
```

```python
wp_decoded = wordpiece_tokenizer.decode(wp_token_ids)

print("Original:")
print(sentence)

print("\nDecoded:")
print(wp_decoded)

print("\nSame text:", sentence.lower() == wp_decoded)
```

```python
comparison = pd.DataFrame({
    "Tokenizer": [
        "GPT",
        "SentencePiece",
        "WordPiece"
    ],
    "Token Count": [
        len(gpt_token_ids),
        len(sp_token_ids),
        len(wp_token_ids)
    ]
})

comparison
```

```python
token_comparison = pd.DataFrame({
    "GPT": pd.Series(gpt_tokens),
    "SentencePiece": pd.Series(sp_tokens),
    "WordPiece": pd.Series(wp_tokens)
})

token_comparison
```

# Observations

## 1. GPT tokenizer

GPT tokenization uses a BPE-based approach. The sentence is divided into subword pieces according to the tokenizer's vocabulary.

## 2. SentencePiece

The SentencePiece-based tokenizer represents the sentence using its own learned vocabulary and segmentation strategy. Word boundaries may be represented using a special marker such as `▁`.

## 3. WordPiece

WordPiece can split words into smaller subword pieces. Continuation pieces are commonly shown using the `##` prefix.

## 4. Token counts

The three tokenizers do not necessarily produce the same number of tokens for the same sentence. This happens because they use different vocabularies and tokenization strategies.

For this experiment:

- GPT produced [7] tokens.
- SentencePiece produced [9] tokens.
- WordPiece produced [14] tokens.

Therefore, the tokenizer with the lowest token count for this particular sentence was GPT.

This result should not be interpreted as meaning that this tokenizer will always produce fewer tokens. Token efficiency depends on the input text and the tokenizer's vocabulary.

## 5. Important conclusion

A token is not necessarily the same thing as a word.

The same word can be represented as one token by one tokenizer and multiple subword tokens by another tokenizer.

Therefore, token count is tokenizer-dependent.
